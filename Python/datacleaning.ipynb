```python
from pathlib import Path
import pandas as pd

output_file = Path("./supply_chain.sql")

# Convert pandas dtype to PostgreSQL datatype
def get_postgres_type(dtype):
    if pd.api.types.is_integer_dtype(dtype):
        return "BIGINT"
    elif pd.api.types.is_float_dtype(dtype):
        return "DOUBLE PRECISION"
    elif pd.api.types.is_datetime64_any_dtype(dtype):
        return "TIMESTAMP"
    else:
        return "TEXT"


with open(output_file, "w", encoding="utf-8") as f:

    # Drop table if it already exists
    f.write("DROP TABLE IF EXISTS supply_chain;\n\n")

    # Create table
    f.write("CREATE TABLE supply_chain (\n")

    column_definitions = []

    for col in df.columns:
        pg_type = get_postgres_type(df[col].dtype)
        column_definitions.append(f'    "{col}" {pg_type}')

    f.write(",\n".join(column_definitions))
    f.write("\n);\n\n")

    # PostgreSQL COPY command
    columns = ", ".join(f'"{col}"' for col in df.columns)

    f.write(
        f"COPY supply_chain ({columns}) "
        "FROM STDIN WITH (FORMAT csv, HEADER true, NULL '\\N');\n"
    )

    # Write the actual data
    df.to_csv(
        f,
        index=False,
        header=True,
        na_rep="\\N"
    )

    # End COPY data
    f.write("\\.\n")


print("SQL file created successfully!")
print("Location:", output_file.resolve())
print("Size:", round(output_file.stat().st_size / (1024 * 1024), 2), "MB")
```

```python
from pathlib import Path

def get_postgres_type(dtype):
    if pd.api.types.is_integer_dtype(dtype):
        return "BIGINT"
    elif pd.api.types.is_float_dtype(dtype):
        return "DOUBLE PRECISION"
    elif pd.api.types.is_datetime64_any_dtype(dtype):
        return "TIMESTAMP"
    else:
        return "TEXT"

sql_file = Path("./create_supply_chain.sql")

with open(sql_file, "w", encoding="utf-8") as f:

    f.write("DROP TABLE IF EXISTS supply_chain;\n\n")

    f.write("CREATE TABLE supply_chain (\n")

    columns = []

    for col in df.columns:
        pg_type = get_postgres_type(df[col].dtype)
        columns.append(f'    "{col}" {pg_type}')

    f.write(",\n".join(columns))
    f.write("\n);\n")

print(sql_file.resolve())
```
